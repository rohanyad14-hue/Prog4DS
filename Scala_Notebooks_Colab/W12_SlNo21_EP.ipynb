## Problem Statement
Create a list of integers from 1 to 10, then use filter to create a new list with only even numbers. Use map to create a new list where each element is multiplied by 3. Use flatMap to create a list of tuples where each integer is paired with its square. Finally, use foldLeft to calculate the product of all elements in the list.

## Objectives
- To master foundational higher-order functional combinators: filter, map, flatMap, and foldLeft.
- To practice declarative, expression-oriented data transformation pipelines without mutable variables.
- To evaluate monad flattening semantics and structural mapping using flatMap.
- To implement recursive accumulation and reduction patterns using foldLeft with custom seed accumulators.
- To contrast the mathematical properties of linear mappings versus associative reductions.

```python
%%writefile W12_SlNo21_FunctionalCombinators.scala
// Program 21: Advanced Functional Combinators in Scala
// Student: Rohan Yadav | Reg. No: 26MML0020

object Experiment21_Combinators {
  def main(args: Array[String]): Unit = {
    val nums: List[Int] = (1 to 10).toList
    println(s"Initial Integer List: $nums")

    // 1. Filter: Extract even numbers
    val evens = nums.filter(x => x % 2 == 0)
    println(s"1. Even Numbers (filter):\n   $evens")

    // 2. Map: Multiply each element by 3
    val tripled = nums.map(x => x * 3)
    println(s"2. Elements Multiplied by 3 (map):\n   $tripled")

    // 3. FlatMap: Pair each integer with its square as tuples
    val pairedWithSquare = nums.flatMap(x => List((x, x * x)))
    println(s"3. Paired with Square as Tuples (flatMap):\n   $pairedWithSquare")

    // 4. FoldLeft: Calculate product of all elements (Factorial of 10)
    val totalProduct = nums.foldLeft(1)((acc, elem) => acc * elem)
    println("=== FoldLeft Product Accumulator Computation ===")
    println(s"Input List: $nums")
    println("Accumulator initial value: 1")
    println("Reduction formula: foldLeft(1)((acc, x) => acc * x)")
    println(s"Final Accumulated Product (10!): $totalProduct")
  }
}
```

## Conclusions & Engineering Insights

The advanced functional combinators experiment showcased the expressive power, conciseness, and declarative nature of Scala's collection API.

- Combinators allow complex transformation pipelines to be expressed without explicit state mutation or index counters.
- The filter combinator isolates subsets cleanly using pure boolean predicates.
- The map combinator guarantees one-to-one structural preservation across collection transformations.
- The flatMap combinator serves as the monadic building block for nested iterations, generators, and collections flattening.
- foldLeft provides safe, tail-recursive aggregations over arbitrary sequences with customizable initial states.

Proficiency in functional combinators is the core skill required for distributed query optimization and resilient dataset manipulation in big data frameworks like Apache Spark and Flink.
